```python
from sklearn.datasets import load_iris

from ML.Ex001 import X_train
```

```python
iris = load_iris()
print(iris)
```

```python
X = iris.data
y = iris.target
```

```python
print(X)
print(X.shape)
```

```python
from sklearn.model_selection import train_test_split
(X_train, X_test, y_train, y_test) = train_test_split(X, y, test_size=0.2, random_state=4)
```

```python
print(X_train.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

```python
y_predict = knn.predict(X_test)
```

```python
print(y_predict)
```

```python
print(y_test)
```

```python
from sklearn import metrics
score = metrics.accuracy_score(y_test, y_predict)
print(score)
```

```python
missing_flower1 = [[3.0, 4.0, 5.0, 2.0]]
print(knn.predict(missing_flower1))
```

```python
classnames = {0:'Setosa', 1: "Versicolor", 2: "Virginica"}
print(classnames[knn.predict(missing_flower1)[0]])
```

```python
from sklearn.metrics import confusion_matrix
result = confusion_matrix(y_test, y_predict)
print(result)
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# 꽃잎 길이와 너비 선택
petal_length = X[:, 2]
petal_width = X[:, 3]

# 품종별 색상 지정
colors = ['red', 'green', 'blue']

for i, color in enumerate(colors):
    plt.scatter(petal_length[y == i], petal_width[y == i], c=color, label=target_names[i])

plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Iris Dataset - Petal Length vs Petal Width')
plt.legend()
plt.show()
```
