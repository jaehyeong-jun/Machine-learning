```python
from pythonjsonlogger.defaults import unknown_default

dachshund_length = [77, 78, 85, 83, 73, 77, 73, 80]
dachshund_height = [25, 28, 29, 30, 21, 22, 17, 35]
samoyed_length = [75, 77, 86, 86, 79, 83, 83, 88]
samoyed_height = [56, 57, 50, 53, 60, 53, 49, 61]
#scatter
import matplotlib.pyplot as plt
plt.scatter(dachshund_length, dachshund_height, c='c', marker= '.')
plt.scatter(samoyed_length, samoyed_height, c='b', marker= "*")
plt.show()
```

```python
#새로운 강아지 발견
unknown_dog_length = [79]
unknown_dog_height = [35]
plt.scatter(dachshund_length, dachshund_height, c='c', marker= '.')
plt.scatter(samoyed_length, samoyed_height, c='b', marker= "*")
plt.scatter(unknown_dog_length, unknown_dog_height, c='r', marker='p')
plt.show()
```

```python
# 닥스 훈트의 레이블 :0
import  numpy as np
dachshund_label = np.zeros(len(dachshund_length))
#사모이드의 레이블 : 1
samoyed_label = np.ones(len(samoyed_length))
print(dachshund_label)
print(samoyed_label)
```

```python
dachshund_date = np.column_stack((dachshund_length, dachshund_height))
print(dachshund_date)
```

```python
samoyed_date = np.column_stack((samoyed_length, samoyed_height))
print(samoyed_date)
```

```python
dogs= np.concatenate((dachshund_date,samoyed_date), axis=0)
labels = np.concatenate((dachshund_label, samoyed_label), axis=0 )
```

```python
print(dogs)
```

```python
print(labels)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(dogs, labels)
unknown_new_dog = [[79, 35]]
print(knn.predict(unknown_new_dog))
```

```python
# 데이터 증강하기
dachshund_length_mean = np.mean(dachshund_length)
dachshund_height_mean = np.mean(dachshund_height)
print(dachshund_length_mean)
print(dachshund_height_mean)
```

```python
samoyed_length_mean = np.mean(samoyed_length)
samoyed_height_mean = np.mean(samoyed_height)
print(samoyed_length_mean)
print(samoyed_height_mean)
```

```python
# 통계적기반으로 데이터를 만들기
new_normal_dachshund_length_date = np.random.normal(dachshund_length_mean, 7, 200)
new_normal_dachshund_height_date = np.random.normal(dachshund_height_mean, 7, 200)

new_normal_samoyed_length_date = np.random.normal(samoyed_length_mean, 7, 200)
new_normal_samoyed_height_date = np.random.normal(samoyed_height_mean, 7, 200)

print(new_normal_dachshund_length_date)
print(new_normal_dachshund_height_date)
print(new_normal_samoyed_length_date)
print(new_normal_samoyed_height_date)
```

```python
# 새롭게 데이터 증강된 그래프 그리기
plt.scatter(new_normal_dachshund_length_date, new_normal_dachshund_height_date, c='b', marker='.')
plt.scatter(new_normal_samoyed_length_date, new_normal_samoyed_height_date, c='m', marker= '*')
plt.show()
```

```python
new_dachshund_date = np.column_stack((new_normal_dachshund_length_date, new_normal_dachshund_height_date))
new_samoyed_date = np.column_stack((new_normal_samoyed_length_date, new_normal_samoyed_height_date))
```

```python
print(new_dachshund_date)
```

```python
print(new_samoyed_date)
```

```python
new_dachshund_label = np.zeros(len(new_dachshund_date))
new_samoyed_label = np.ones(len(new_samoyed_date))
print(new_dachshund_label)
print(new_samoyed_label)
```

```python
new_dogs = np.concatenate((new_dachshund_date, new_samoyed_date))
new_labels = np.concatenate((new_dachshund_label, new_samoyed_label))
```

```python
print(new_dogs)
```

```python
print(new_labels)
```

```python
from sklearn.model_selection import train_test_split
(X_train, X_test, y_train,y_test) = train_test_split(new_dogs, new_labels, test_size=0.2, random_state=0)
```

```python
print(X_train.shape)
```

```python
k = 5
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)
print(f'훈련도 정확도: {knn.score(X_train, y_train)}')
```

```python
y_predict = knn.predict(X_test)
print(y_predict)
```

```python
print(y_test)
```

```python
from sklearn.metrics import accuracy_score
print(f'테스트 정확도 : {accuracy_score(y_test, y_predict)}')
```
